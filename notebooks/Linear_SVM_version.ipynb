# *Importing Necessary Libraries and functions*

```
import pandas as pd
from math import log2
from math import pi
from math import exp
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import GridSearchCV
from sklearn import svm
```

# *Reading Data*

```
df = pd.read_csv('/content/data.csv')
```

# *Drop Bad Data*

```
df.dropna()
df.isnull().sum()
```

# *Drop Unwanted Data*

```
df.drop('id',axis=1,inplace=True)
df.drop('Unnamed: 32',axis=1,inplace=True)
```

# *Convert Malignant diagnosis into 1 and Benign into 0*

```
df['diagnosis'].astype(str)
df['diagnosis'] = [1 if val == 'M' else 0 for val in df['diagnosis'].values ]
```

# *Select Data and Labels*

```
x = df.iloc[:,1:len(df.columns)]
y = df['diagnosis'].values
```

# *Data scaling and normalization*

```
scaler = StandardScaler()
x = scaler.fit_transform(x)
```

# *Declaring C parameter list*

```
C=[0.1, 1, 10, 100, 1000,10000]
```

# *Looping in C list and calculating MDL without fitting into Gaussian*
\begin{equation}
\frac{1}{2} N_{E}\log_{2}(N) -\sum_{i=0}^{N-1}\log_{2}(P(x_{i}))
\end{equation}


```
print("-------------------------------------------------------------------------")
for i in C:
  svc = svm.SVC(kernel='linear', C=i).fit(x, y)
  dec = svc.decision_function(x)
  w_norm = np.linalg.norm(svc.coef_)#norm of the hyperplane
  N= x.shape[0]
  dist = dec / w_norm #Distance calculation formula when output of decision function and w_norm is given.
  right_term = np.sum(np.log2(abs(dist))) #implementing Right term of the equation
  N_e = svc.support_vectors_.shape[0] #Number of support vectors
  MDL = (1/2)*N_e*log2(N) -right_parameter
  print("Number of support vectors: ",N_e)
  print("Number of training points: ",N)
  print("Right Term of the final MDL equation: ",right_term)
  print("MDL: ",MDL)
  print("---------------------------------------------------------------------------")
```

# *Looping in C list and calculating MDL with fitting into Gaussian*
\begin{equation}
\frac{1}{2} N_{E}\log_{2}(N)  + \frac{1}{2} N\log_{2}(2\pi \sigma ^{2}e)
\end{equation}

```
print("-------------------------------------------------------------------------")
for i in C:
  svc = svm.SVC(kernel='linear', C=i).fit(x, y)
  dec = svc.decision_function(x)
  w_norm = np.linalg.norm(svc.coef_)
  N= x.shape[0]
  dist = dec / w_norm
  var = np.var(dist)#Variance of DIstance
  N_e = svc.support_vectors_.shape[0]
  MDL = (1/2)*N_e*log2(N) + (1/2)*N*log2(2*pi*exp(1)*var)
  print("Number of support vectors: ",N_e)
  print("Number of training points: ",N)
  print("Variance: ",var)
  print("MDL: ",MDL)
  print("---------------------------------------------------------------------------")
```

# *Original Grid Search with cross validation*

```
param_grid = {'C': [0.1, 1, 10, 100, 1000,10000],
              'kernel': ['linear']}
grid = GridSearchCV(svm.SVC(), param_grid, refit = True, verbose = 0)
grid.fit(x, y)
print("Best hyperparameter for grid search: ", grid.best_params_)
```
